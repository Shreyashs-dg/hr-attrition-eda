## Data Understanding — Summary

- Dataset contains 1470 rows and 35 columns, no missing values and no duplicate rows.
- Target variable `Attrition` is imbalanced: ~83.9% "No" vs ~16.1% "Yes" — this will need 
  to be accounted for in later modeling (Phase 2), and is worth calling out now during EDA.
- Three constant columns carry no information and will be dropped: `EmployeeCount`, 
  `Over18`, `StandardHours` (all have only one unique value across the dataset).
- `EmployeeNumber` is a unique identifier per employee (1470 unique values for 1470 rows) 
  and will be excluded from analysis — it's an ID, not a feature.
- Columns were grouped into categorical (text-based), continuous numeric, and ordinal 
  (rating-scale) types to guide appropriate visualizations in later modules.
- Overall, the dataset is clean with no missing-value or duplicate handling required — 
  cleaning in Module 3 will focus on dropping useless columns and setting correct dtypes.
