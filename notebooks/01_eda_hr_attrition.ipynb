## Data Understanding — Summary

- Dataset contains 1470 rows and 35 columns, no missing values and no duplicate rows.
- Target variable `Attrition` is imbalanced: ~83.9% "No" vs ~16.1% "Yes" — this will need 
  to be accounted for in later modeling (Phase 2), and is worth calling out now during EDA.
- Three constant columns carry no information and will be dropped: `EmployeeCount`, 
  `Over18`, `StandardHours` (all have only one unique value across the dataset).
- `EmployeeNumber` is a unique identifier per employee (1470 unique values for 1470 rows) 
  and will be excluded from analysis — it's an ID, not a feature.
- Columns were grouped into categorical (text-based), continuous numeric, and ordinal 
  (rating-scale) types to guide appropriate visualizations in later modules.
- Overall, the dataset is clean with no missing-value or duplicate handling required — 
  cleaning in Module 3 will focus on dropping useless columns and setting correct dtypes.

## Univariate Analysis — Summary

- **Skewed columns:** `YearsSinceLastPromotion` (skew ≈ 1.98) and `YearsAtCompany` 
  (skew ≈ 1.76) are the most right-skewed — most employees were promoted recently 
  or have shorter tenure, with a long tail of longer-tenured outliers. `MonthlyIncome` 
  (skew ≈ 1.37) and `TotalWorkingYears` (skew ≈ 1.12) are moderately right-skewed too, 
  typical for income/tenure data.
- **Roughly normal/symmetric columns:** `Age` (skew ≈ 0.41), `MonthlyRate`, `DailyRate`, 
  and `HourlyRate` (skew close to 0) show fairly balanced, symmetric distributions.
- **Columns with visible outliers:** `TrainingTimesLastYear` has the most outlier points 
  (~16% by IQR), followed by `MonthlyIncome` (~7.8%), `YearsSinceLastPromotion` (~7.3%), 
  and `YearsAtCompany` (~7.1%). These are mostly high-tenure or high-income employees 
  at the upper end, not obvious data errors.
- **Notable single-column patterns:**
  - About 74.6% of employees have 2 years or fewer since their last promotion — 
    promotions cluster heavily toward the recent end.
  - `OverTime` is imbalanced — 71.7% of employees do **not** work overtime, only 
    28.3% do.
  - `PerformanceRating` only contains two values in this dataset — **3 and 4** — 
    despite the field theoretically supporting a 1-4 scale. No employee is rated 
    1 or 2, which limits how useful this column will be for differentiating 
    performance later.

## Bivariate / Multivariate Analysis — Summary

- Employees who left the company skew younger (median age 32 vs 36), earn less 
  (median MonthlyIncome ~3,202 vs ~5,204), and have shorter tenure and experience 
  (median YearsAtCompany 3 vs 6, TotalWorkingYears 7 vs 10) compared to those who stayed.
- **OverTime is the strongest single behavioral driver of attrition** — employees who 
  work overtime leave at ~30.5%, nearly three times the ~10.4% rate for those who don't.
- Attrition also varies by group: Sales (~20.6%) and HR (~19.0%) have higher attrition 
  than R&D (~13.8%); Single employees leave at ~25.5% vs ~12.5% (Married) and ~10.1% 
  (Divorced); frequent travelers leave at ~24.9% vs ~8.0% for non-travelers.
- All four satisfaction/involvement scores (JobSatisfaction, EnvironmentSatisfaction, 
  WorkLifeBalance, JobInvolvement) show a consistent downward trend — attrition drops 
  as the score rises, with JobInvolvement showing the sharpest gap (~33.7% at score 1 
  vs ~9.0% at score 4).
- The correlation heatmap confirms these patterns numerically: TotalWorkingYears, 
  JobLevel, YearsInCurrentRole, MonthlyIncome, and Age show the strongest negative 
  correlations with attrition (roughly -0.16 to -0.17), meaning more seniority/tenure/
  income associates with lower attrition risk. No numeric feature shows a strong 
  positive correlation.
- **Attrition is highly uneven across job roles** — Sales Representative has by far the 
  highest rate (~39.8%), followed by Laboratory Technician and HR (~24%), while 
  Research Director and Manager show the lowest (~2.5% and ~4.9%).
- A two-factor check shows overtime and low job satisfaction compound each other: 
  employees working overtime with JobSatisfaction scores 1-3 show attrition rates of 
  34-38%, versus just 7-10% for non-overtime employees regardless of satisfaction — 
  overtime appears to be the dominant risk factor, made worse by low satisfaction.
